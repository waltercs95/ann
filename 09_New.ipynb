09 . Write a python program to illustrate ART neural network: Clustering 2D Data Using ART1-  
a. Implement the ART1 algorithm for binary input data.  
b. Given a set of binary vectors, cluster them into categories based on similarity.  
c. Use vigilance parameter (ρ) to control cluster formation. Clustering 2D Data using Winner-Takes-All (WTA)  
Network Implement a Winner-Takes-All (WTA) neural network to cluster 2D data points. 
d. Each neuron represents a cluster center, and only the winning neuron (closest to input) updates its weights.  
e. Train the network on randomly generated 2D points and visualize the clusters.  

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([
    [0, 1, 0, 1],
    [0, 0, 1, 1],
    [1, 0, 1, 1],
    [1, 0, 0, 0],
    [1, 0, 0, 1]
])

rho = 0.7
weights = []

for x in X :
    assigned = False

    for i in range(len(weights)) :
        match = np.sum(np.minimum(x, weights[i]))/ np.sum(x)
    
        if match >= rho :
            weights[i] = np.minimum(x, weights[i])
            assigned = True
            break

    if not assigned :
        weights.append(x.copy())

print("ART-I Clusters :")
for i, w in enumerate(weights) :
    print("Cluster", i+1, ":", w)

np.random.seed(42)

X = np.vstack([
    np.random.randn(50, 2) + [2, 2],
    np.random.randn(50, 2) + [7,7],
    np.random.randn(50, 2) + [2, 7]
])

k = 3
alpha = 0.1
centers = X[np.random.choice(len(X), k, replace=False)]

for epoch in range(50) :
    for x in X :
        distance = np.linalg.norm(centers - x, axis=1)
        winner = np.argmin(distance)
        centers[winner] += alpha * (x - centers[winner])

labels = []
for x in X :
    distance = np.linalg.norm(centers - x, axis=1)
    labels.append(np.argmin(distance))
labels = np.array(labels)

print("\nWTA Cluster Centers :")
print(centers)

plt.scatter(
    X[:,0], X[:,1],
    c=labels, cmap='viridis'
)

plt.scatter(
    centers[:,0],
    centers[:,1],
    color="red",
    marker="X",
    s=200,
    label="Cluster Centers"
)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("WTA Clustering")
plt.legend()
plt.show()
```

1. What is ART?

ART stands for Adaptive Resonance Theory. It is a neural network used for unsupervised learning and clustering.

2. What is the main purpose of ART1?

To group similar binary input patterns into different categories.

3. What is the vigilance parameter (ρ)?

It determines the minimum similarity required for an input pattern to join an existing cluster.

4. What happens when vigilance is high?

More clusters are generally formed because the similarity requirement is strict.

5. What happens when vigilance is low?

Fewer clusters are generally formed because less similarity is required.

6. What is WTA?

WTA stands for Winner-Takes-All. It is a competitive learning neural network used for clustering.

7. What is the main purpose of WTA?

To group similar data points into clusters based on distance.
